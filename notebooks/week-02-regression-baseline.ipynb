### Learning goals

- Separate features from a numeric target.
- Build a mean baseline before fitting a model.
- Compare predictions with MAE and RMSE.


## 1. Setup


```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.model_selection import train_test_split
```

## 2. Create a small reproducible dataset

The feature is weekly practice time. The target is a later assessment score. The random noise keeps the relationship realistic while remaining reproducible.


```python
rng = np.random.default_rng(42)

practice_hours = np.linspace(1, 12, 60)
scores = 48 + 3.4 * practice_hours + rng.normal(0, 4.5, size=practice_hours.size)
scores = np.clip(scores, 0, 100)

X = practice_hours.reshape(-1, 1)
y = scores

print("feature shape:", X.shape)
print("target shape:", y.shape)
```

## 3. Split before training

The test set represents new examples that the model did not use for fitting.


```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

print("training rows:", len(X_train))
print("test rows:", len(X_test))
```

## 4. Build a baseline

The baseline predicts the training-set mean for every test row.


```python
baseline = DummyRegressor(strategy="mean")
baseline.fit(X_train, y_train)
baseline_predictions = baseline.predict(X_test)

baseline_mae = mean_absolute_error(y_test, baseline_predictions)
baseline_rmse = mean_squared_error(y_test, baseline_predictions) ** 0.5

print(f"Baseline MAE: {baseline_mae:.2f}")
print(f"Baseline RMSE: {baseline_rmse:.2f}")
```

## 5. Fit linear regression


```python
model = LinearRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)

model_mae = mean_absolute_error(y_test, predictions)
model_rmse = mean_squared_error(y_test, predictions) ** 0.5

print(f"Linear regression MAE: {model_mae:.2f}")
print(f"Linear regression RMSE: {model_rmse:.2f}")
print(f"Slope: {model.coef_[0]:.2f}")
print(f"Intercept: {model.intercept_:.2f}")
```

## 6. Compare the models


```python
improvement = baseline_mae - model_mae
print(f"MAE improvement over baseline: {improvement:.2f} points")

labels = ["Mean baseline", "Linear regression"]
values = [baseline_mae, model_mae]

plt.figure(figsize=(7, 4))
plt.bar(labels, values, color=["#9AA9B5", "#147D88"])
plt.ylabel("Mean absolute error")
plt.title("Lower MAE is better")
plt.show()
```

## 7. Inspect the fitted line


```python
order = np.argsort(X_test[:, 0])

plt.figure(figsize=(8, 4.5))
plt.scatter(X_train[:, 0], y_train, alpha=0.55, label="Training data")
plt.scatter(X_test[:, 0], y_test, color="#E07A5F", label="Test data")
plt.plot(X_test[order, 0], predictions[order], color="#147D88", linewidth=3, label="Fitted line")
plt.xlabel("Practice hours per week")
plt.ylabel("Assessment score")
plt.title("Linear regression on held-out examples")
plt.legend()
plt.show()
```
