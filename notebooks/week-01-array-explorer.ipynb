## 2. Create two arrays

The two arrays line up by position. The first practice time matches the first score, and so on.


```python
hours = np.array([1, 2, 3, 4, 5, 6])
scores = np.array([52, 57, 63, 68, 74, 79])

print("hours:", hours)
print("scores:", scores)
print("hours shape:", hours.shape)
print("scores data type:", scores.dtype)
```

## 3. Summarize the data

NumPy can calculate statistics across an entire array without a Python loop.


```python
print("mean score:", scores.mean())
print("lowest score:", scores.min())
print("highest score:", scores.max())
print("score range:", scores.max() - scores.min())
```

## 4. Filter with a Boolean mask

A mask stores one True or False value for every item. Use it to select matching values.


```python
passing_mask = scores >= 70

print("mask:", passing_mask)
print("scores at or above 70:", scores[passing_mask])
print("matching practice hours:", hours[passing_mask])
```

## 6. Your first machine-learning model

We will begin with an ordinary Python dictionary so the data is easy to read. Every song maps to a danceability rating, an energy rating, and a mood. The ratings are hand-created for teaching; they are not official Spotify measurements.

Next, we will pull those values into NumPy arrays. Then k-nearest neighbors will find nearby songs and use their most common mood. scikit-learn handles the distance calculations.

The entire machine-learning step is only three lines: **create, fit, predict**.


```python
song_data = {
    "Ocean Eyes": {"danceability": 2, "energy": 3, "mood": "chill"},
    "The Scientist": {"danceability": 3, "energy": 2, "mood": "chill"},
    "Someone Like You": {"danceability": 4, "energy": 3, "mood": "chill"},
    "Uptown Funk": {"danceability": 8, "energy": 9, "mood": "energetic"},
    "Shake It Off": {"danceability": 8, "energy": 8, "mood": "energetic"},
    "Levitating": {"danceability": 9, "energy": 8, "mood": "energetic"},
}

for song_name, details in song_data.items():
    print(
        f"{song_name:18} | danceability={details['danceability']} "
        f"| energy={details['energy']} | mood={details['mood']}"
    )
```

### Turn the dictionary into NumPy arrays

A model needs a table of numbers and a matching list of answers. This loop pulls both out of the readable dictionary. Each feature row becomes `[danceability, energy]`.


```python
song_names = []
feature_rows = []
mood_labels = []

for song_name, details in song_data.items():
    song_names.append(song_name)
    feature_rows.append([details["danceability"], details["energy"]])
    mood_labels.append(details["mood"])

song_names = np.array(song_names)
song_features = np.array(feature_rows)
moods = np.array(mood_labels)

print("features:")
print(song_features)
print("moods:", moods)
```

### Create, fit, predict

Choose a song by name. Its name looks up the two ratings; the model predicts from those numbers, not from the title itself. The three highlighted lines are the whole ML workflow.


```python
songs_to_predict = {
    "Blinding Lights": [8, 9],
    "drivers license": [3, 4],
    "As It Was": [7, 6],
}

song_to_predict = "Blinding Lights"
new_song = np.array([songs_to_predict[song_to_predict]])

model = KNeighborsClassifier(n_neighbors=3)   # create
model.fit(song_features, moods)               # fit
predicted_mood = model.predict(new_song)[0]   # predict

print(f"Prediction for {song_to_predict}: {predicted_mood}")
```

```python
palette = {"chill": "#147D88", "energetic": "#E07A5F"}

plt.figure(figsize=(10, 6.5))
for mood in ["chill", "energetic"]:
    matching_songs = moods == mood
    plt.scatter(
        song_features[matching_songs, 0],
        song_features[matching_songs, 1],
        label=mood,
        color=palette[mood],
        s=90,
    )

for index, name in enumerate(song_names):
    plt.annotate(
        name,
        (song_features[index, 0], song_features[index, 1]),
        xytext=(6, 6),
        textcoords="offset points",
        fontsize=8,
    )

plt.scatter(
    new_song[0, 0],
    new_song[0, 1],
    marker="*",
    s=320,
    color="#D5A12E",
    edgecolor="black",
    label=f"{song_to_predict} → {predicted_mood}",
)
plt.xlabel("Danceability (1–10)")
plt.ylabel("Energy (1–10)")
plt.title("Predict a named song from nearby examples")
plt.xlim(1, 10)
plt.ylim(1, 10)
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```
